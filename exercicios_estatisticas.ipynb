```python
import pandas as pd
```

```python
import kagglehub
caminho = kagglehub.dataset_download("equilibriumm/sleep-efficiency")

print("Caminho do dataset: ", caminho)
```

```python
path_full = caminho + "/Sleep_Efficiency.csv"
data = pd.read_csv(path_full)
data
```

**Algumas atividades para treinar**

Média simples

```python
duracao_sono = data['Sleep duration']

print(f'Média Simples: {duracao_sono.mean()}')
print('Média de sono de 7h e 30min aproximadamente de duração de sono')
```

Média Truncada

```python
from scipy.stats import trim_mean

media_truncada = trim_mean(duracao_sono, 0.1)

print(f'Média truncada: {media_truncada}')
```

Média ponderada

```python
data = {
    "Aluno": ["Ana", "Sabão", "Sakura", "Iruka", "Goku"],
    "Nota": [8.0, 7.5, 9.0, 6.5, 8.5],
    "Peso": [2, 3, 1, 4, 2],
}

df = pd.DataFrame(data)
```

```python
media_ponderada = (df['Nota'] * df['Peso']).sum() / df['Peso'].sum()
print(f'Média ponderada: {media_ponderada}:' )
```
